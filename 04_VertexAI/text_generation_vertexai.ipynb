### Generation Configuration

You can control the creativity and randomness of text generation using these parameters:

- **Temperature (0.0-1.0)**: Controls randomness. Higher = more creative, lower = more deterministic
- **Top-p (0.0-1.0)**: Nucleus sampling. Controls diversity by considering only top tokens
- **Top-k**: Limits vocabulary to top k tokens
- **Max output tokens**: Maximum length of generated text


```python
from vertexai.generative_models import GenerationConfig

# Creative configuration (high temperature)
creative_config = GenerationConfig(
    temperature=0.9,  # High creativity
    top_p=0.95,
    top_k=40,
    max_output_tokens=8192
)

# Deterministic configuration (low temperature)
deterministic_config = GenerationConfig(
    temperature=0.1,  # Low creativity, more consistent
    top_p=0.8,
    top_k=20,
    max_output_tokens=8192
)

# Compare different configurations
prompt = "Write a haiku about the feeling of discovering a perfect book that changes your life"

print("=== Creative Response (Temperature=0.9) ===")
response = model.generate_content(prompt, generation_config=creative_config)
print(response.text)

print("\n=== Deterministic Response (Temperature=0.1) ===")
response = model.generate_content(prompt, generation_config=deterministic_config)
print(response.text)
```

## 2. Text Embeddings

Text embeddings convert text into numerical vectors that capture semantic meaning. These vectors enable:

- **Semantic Search**: Find similar content based on meaning
- **Clustering**: Group related texts together
- **Classification**: Categorize text by content
- **Recommendation Systems**: Suggest similar content
- **Anomaly Detection**: Identify outliers in text data

### Different Task Types

Embeddings can be optimized for different tasks:
- `RETRIEVAL_QUERY` - For search queries
- `RETRIEVAL_DOCUMENT` - For documents to be searched
- `SEMANTIC_SIMILARITY` - For comparing text similarity
- `CLASSIFICATION` - For categorizing text
- `CLUSTERING` - For grouping related texts

```python
from vertexai.language_models import TextEmbeddingModel, TextEmbeddingInput

# Initialize the embedding model
embedding_model = TextEmbeddingModel.from_pretrained('text-multilingual-embedding-002')

# Example texts for different literary themes
texts = [
    "A coming-of-age story about a young wizard discovering his magical powers",
    "An epic fantasy quest to destroy an ancient evil artifact",
    "A romantic tale of two lovers from feuding families",
    "A passionate love story set against the backdrop of war",
    "A psychological thriller about a detective hunting a serial killer",
    "A gripping mystery involving a locked-room murder"
]

# Generate embeddings
embeddings = embedding_model.get_embeddings(texts)

print(f"Generated {len(embeddings)} embeddings")
print(f"Each embedding has {len(embeddings[0].values)} dimensions")
print(f"First embedding (first 5 values): {embeddings[0].values[:5]}")
```

```python
import numpy as np

# Calculate semantic similarity using cosine similarity
def cosine_similarity(vec1, vec2):
    """Calculate cosine similarity between two vectors"""
    dot_product = np.dot(vec1, vec2)
    norm_a = np.linalg.norm(vec1)
    norm_b = np.linalg.norm(vec2)
    return dot_product / (norm_a * norm_b)

# Calculate similarities between all pairs
print("=== Semantic Similarity Matrix ===")
print("Texts:")
for i, text in enumerate(texts):
    print(f"{i}: {text}")

print("\nSimilarity Matrix:")
print("     ", end="")
for i in range(len(texts)):
    print(f"{i:8}", end="")
print()

for i in range(len(texts)):
    print(f"{i}: ", end="")
    for j in range(len(texts)):
        sim = cosine_similarity(embeddings[i].values, embeddings[j].values)
        print(f"{sim:.4f}  ", end="")
    print()

# Find most similar pairs
print("\n=== Most Similar Text Pairs ===")
similarities = []
for i in range(len(texts)):
    for j in range(i+1, len(texts)):
        sim = cosine_similarity(embeddings[i].values, embeddings[j].values)
        similarities.append((sim, i, j))

similarities.sort(reverse=True)
for sim, i, j in similarities[:3]:
    print(f"Similarity: {sim:.4f}")
    print(f"  Text {i}: {texts[i]}")
    print(f"  Text {j}: {texts[j]}")
    print()
```

```python
# Demonstrate different task types
print("=== Different Task Types ===")

# Example: Search query vs document
query = "I want a book that will make me cry happy tears"
documents = [
    "A heartwarming story about an elderly man who finds love again after losing his wife",
    "A tale of friendship between a lonely child and a magical creature",
    "The inspiring journey of a community coming together to save their local library",
    "A moving memoir about overcoming adversity and finding hope in dark times"
]

# Create embeddings with different task types
query_embedding = embedding_model.get_embeddings([
    TextEmbeddingInput(text=query, task_type="RETRIEVAL_QUERY")
])

document_embeddings = embedding_model.get_embeddings([
    TextEmbeddingInput(text=doc, task_type="RETRIEVAL_DOCUMENT") for doc in documents
])

print(f"Query: {query}")
print("\nDocument Similarities:")
for i, doc in enumerate(documents):
    similarity = cosine_similarity(query_embedding[0].values, document_embeddings[i].values)
    print(f"{similarity:.4f}: {doc}")

print("\n" + "="*50)
```

```python
# Example: Text Classification with Embeddings
print("=== Text Classification Example ===")

# Sample book reviews from readers
reviews = [
    "This book completely changed my perspective on life. I couldn't put it down!",
    "Boring plot, predictable ending. I regret buying this.",
    "Beautiful writing style, though the story dragged in places.",
    "Absolutely terrible characters and confusing storyline. Waste of time.",
    "A masterpiece! Every chapter left me wanting more."
]

# Classification categories (represented as text)
categories = [
    "positive book review",
    "negative book review"
]

# Get embeddings for classification task
review_embeddings = embedding_model.get_embeddings([
    TextEmbeddingInput(text=review, task_type="CLASSIFICATION") for review in reviews
])

category_embeddings = embedding_model.get_embeddings([
    TextEmbeddingInput(text=cat, task_type="CLASSIFICATION") for cat in categories
])

print("Review Classification Results:")
for i, review in enumerate(reviews):
    similarities = []
    for j, category in enumerate(categories):
        sim = cosine_similarity(review_embeddings[i].values, category_embeddings[j].values)
        similarities.append((sim, category))
    
    best_category = max(similarities, key=lambda x: x[0])
    print(f"\nReview: {review}")
    print(f"Classification: {best_category[1]} (confidence: {best_category[0]:.4f})")

print("\n" + "="*50)
```

```python
# Example: Simple Clustering with Embeddings
print("=== Text Clustering Example ===")

# Mixed book descriptions from different genres
mixed_texts = [
    "A space opera about rebels fighting against an intergalactic empire",
    "A cozy mystery set in a small English village with a amateur detective",
    "A dystopian novel about a society where emotions are forbidden",
    "A romance between a baker and a food critic in modern-day Paris",
    "A hard science fiction story about colonizing Mars",
    "A detective novel featuring a brilliant but troubled investigator",
    "A fantasy epic with dragons, magic, and ancient prophecies",
    "A contemporary love story about second chances and healing"
]

# Get embeddings for clustering
cluster_embeddings = embedding_model.get_embeddings([
    TextEmbeddingInput(text=text, task_type="CLUSTERING") for text in mixed_texts
])

# Simple clustering: group by highest similarity
print("Clustering Results:")
print("(Books are grouped by similar genres/themes)")
print()

used_indices = set()
clusters = []

for i in range(len(mixed_texts)):
    if i in used_indices:
        continue
    
    cluster = [i]
    used_indices.add(i)
    
    for j in range(i+1, len(mixed_texts)):
        if j in used_indices:
            continue
        
        sim = cosine_similarity(cluster_embeddings[i].values, cluster_embeddings[j].values)
        if sim > 0.65:  # Lowered threshold for book clustering
            cluster.append(j)
            used_indices.add(j)
    
    if cluster:
        clusters.append(cluster)

# Display clusters
for i, cluster in enumerate(clusters):
    print(f"Genre Cluster {i+1}:")
    for idx in cluster:
        print(f"  - {mixed_texts[idx]}")
    print()

print("Note: Books in the same cluster share similar themes or genres")
```
